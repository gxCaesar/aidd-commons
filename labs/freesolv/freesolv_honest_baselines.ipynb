# FreeSolv v0.52 honest baselines

This paired notebook reads the reviewed aggregate `reviewed-results.json` artifact. It does not fit models, change splits, download data, or perform another test evaluation.

## Reading order and caveat

The scaffold split is the primary result because it is the declared harder generalization contrast. The random split is shown second as an easier contrast. The displayed 90% intervals use a fixed validation-calibrated radius; their empirical coverage is marginal coverage, not conditional calibration.

For a fresh run, first complete the data and pre-test workflow documented in `README.md`. This notebook is intentionally result-reading only.

```python
from __future__ import annotations

import json
from pathlib import Path

results_path = Path("labs/freesolv/reviewed-results.json")
results = json.loads(results_path.read_text(encoding="utf-8"))
assert results["artifact_type"] == "reviewed-aggregate-results"
assert results["primary_split"] == "scaffold"


def print_split(label: str, split: dict[str, object]) -> None:
    print(label)
    point_metrics = split["point_metrics"]
    interval_metrics = split["interval_metrics"]
    for model_name in ("train_mean", "morgan_ridge", "morgan_random_forest"):
        values = point_metrics[model_name]
        print(f"  {model_name}: MAE={values['mae']:.6f}, RMSE={values['rmse']:.6f}")
    for model_name in ("morgan_ridge", "morgan_random_forest"):
        values = interval_metrics[model_name]
        print(
            f"  {model_name} interval: coverage={values['coverage']:.6f}, "
            f"mean width={values['mean_width']:.6f}"
        )


print_split("Scaffold-primary results", results["splits"]["scaffold"])
print()
print_split("Random-split contrast", results["splits"]["random"])
print(
    "\nInterval caveat: empirical intervals report marginal coverage, not conditional calibration."
)
```
